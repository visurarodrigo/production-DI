# Prediction - Using the Saved Model on New Input

Loads the trained Gradient Boosting model and makes a prediction for a single new production scenario.

## 1. Imports


```python
from sklearn.preprocessing import StandardScaler
import joblib
import pandas as pd
```

## 2. Load Model & Feature Scaler 


```python
model = joblib.load("gradient_boosting_model.joblib")
X_scaler = joblib.load("X_scaler.joblib")
y_scaler = joblib.load("y_scaler.joblib")
```

## 3. Define a New Input Scenario
A single hypothetical production week, built manually with the same encoded feature structure as the training data.


```python
# REAL INPUT VALUES

planned_qty_real = 35
no_of_workers_real = 128

# CREATE INPUT DATA

new_data = pd.DataFrame({
    "grade": [1],

    "planned_qty": [planned_qty_real],
    "no_of_workers": [no_of_workers_real],

    "bottleneck_flag": [0],

    "order_category_Others": [1],
    "order_category_School": [0],

    "factory_unit_Factory Unit 02": [1],
    "product_family_Ladies": [1],

    "financial_month_Aug": [0],
    "financial_month_Dec": [0],
    "financial_month_Feb": [0],
    "financial_month_Jan": [1],
    "financial_month_Jul": [0],
    "financial_month_Jun": [0],
    "financial_month_Mar": [0],
    "financial_month_May": [0],
    "financial_month_Nov": [0],
    "financial_month_Oct": [0],
    "financial_month_Sep": [0],

    "week_of_month_2": [1],
    "week_of_month_3": [0],
    "week_of_month_4": [0],
    "week_of_month_5": [0],

    "size_id_7": [1],
    "size_id_8": [0],
    "size_id_9": [0],
    "size_id_10": [0],
    "size_id_11": [0],
    "size_id_12": [0],
    "size_id_13": [0]
})
```

```python
new_data
```

## 4. Standardize Numerical Inputs

The numerical features are transformed using the previously fitted `X_scaler` to match the scaling applied during model training.

```python
# STANDARDIZE NUMERICAL INPUTS

values_to_scale = new_data[
    ["planned_qty", "no_of_workers"]
]

scaled_values = X_scaler.transform(values_to_scale)

new_data[
    ["planned_qty", "no_of_workers"]
] = scaled_values
```

```python
new_data
```

## 5. Align Column Order
Matches the exact column order the model was trained on. 


```python
# MAKE SURE COLUMN ORDER IS CORRECT

expected_columns = [
    "grade",
    "planned_qty", 
    "no_of_workers",
    "bottleneck_flag",
    "order_category_Others",
    "order_category_School",
    "factory_unit_Factory Unit 02",
    "product_family_Ladies",
    "financial_month_Aug",
    "financial_month_Dec",
    "financial_month_Feb",
    "financial_month_Jan",
    "financial_month_Jul",
    "financial_month_Jun",
    "financial_month_Mar",
    "financial_month_May",
    "financial_month_Nov",
    "financial_month_Oct",
    "financial_month_Sep",
    "week_of_month_2",
    "week_of_month_3",
    "week_of_month_4",
    "week_of_month_5",
    "size_id_7",
    "size_id_8",
    "size_id_9",
    "size_id_10",
    "size_id_11",
    "size_id_12",
    "size_id_13"
]

new_data = new_data[expected_columns]
```

## 6. Predict 



```python
# MAKE PREDICTION

prediction_scaled = model.predict(new_data)
      
# CONVERT PREDICTION BACK TO REAL VALUE

prediction_real = y_scaler.inverse_transform(
    prediction_scaled.reshape(-1, 1)
)

# DISPLAY RESULTS

print("PREDICTION")

print("Real planned quantity :", planned_qty_real)
print("Real number of workers:", no_of_workers_real)

print("\nStandardized planned_qty:",
      new_data["planned_qty"].iloc[0])

print("Standardized no_of_workers:",
      new_data["no_of_workers"].iloc[0])

print("\nPredicted actual_qty:",
      prediction_real[0][0])
```
