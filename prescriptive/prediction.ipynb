```python
import pandas as pd
import joblib

def predict_required_hours(planning_data, model_path='resource_planning_model_hours.joblib'):
    """
    Loads the saved model and predicts working_hours needed for a specific production target.
    """
    # 1. Load the saved pipeline
    pipeline = joblib.load(model_path)
    
    # 2. Apply the exact same feature engineering
    df_pred = planning_data.copy()
    df_pred['machine_worker_ratio'] = df_pred['machine_count'] / df_pred['no_of_workers']
    
    # 3. Predict
    predictions = pipeline.predict(df_pred)
    
    # Return rounded to 2 decimal places for practical hours tracking
    return predictions.round(2)

# --- EXAMPLE USAGE ---
if __name__ == "__main__":
    # Scenario: "I want to produce exactly 50 shoes (actual_qty). 
    # I have 15 workers and 4 machines. How many hours will it take?"
    new_scenario = pd.DataFrame([{
        'financial_month': 'May',
        'week_no': 2,
        'week_of_month': 2,
        'factory_unit': 'Factory Unit 01',
        'product_family': 'Ladies',
        'item_name': 'BINOLI (VAST) - BROWN',
        'size_id': 10,
        'grade': 'A',
        'rate': 4500.00,
        'actual_qty': 50,  # <-- This is your target production goal
        'no_of_workers': 15,
        'total_duration_per_shoe_mins': 120.0,
        'machine_count': 4,
        'order_category': 'Others'
    }])

    # Get the prediction
    predicted_hours = predict_required_hours(new_scenario)
    
    print(f"To produce {new_scenario['actual_qty'].iloc[0]} shoes with {new_scenario['no_of_workers'].iloc[0]} workers:")
    print(f"Estimated Required Working Hours: {predicted_hours[0]} hours")
```
